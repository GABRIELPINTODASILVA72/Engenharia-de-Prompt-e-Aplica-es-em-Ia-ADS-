#**MISSÃO 01** AUTOMAÇÃO DE ARQUIVOS


```python
'''
Criar um programa que organize arquivos bagunçados em pastas automaticamente,
baseando-se apenas na extensão do arquino (ex: .pdf, .jpg). baseando nos
aquivos de exemplos adicione aquivos para testar mas com que nao interfira
no codigo que depois apenas eu ou vc tire esses exemplos. obs: nao tire essa
mensagem apenas adicine o codigo abaixo. por favor coloque # explicando a linha


'''

import os # Importa o módulo os para interagir com o sistema operacional (caminhos, diretórios)
import shutil # Importa o módulo shutil para operações de alto nível em arquivos e diretórios (como mover)

def criar_arquivos_teste(diretorio='.'):
    """Cria arquivos de teste com diferentes extensões.""" # Docstring explicando a função
    # Imprime o diretório onde os arquivos de teste serão criados
    print(f"Criando arquivos de teste no diretório: {os.path.abspath(diretorio)}")
    # Lista de extensões que serão usadas para os arquivos de teste
    extensoes = ['.txt', '.pdf', '.jpg', '.png', '.doc', '.mp3']
    # Loop para criar 10 arquivos de teste
    for i in range(1, 11):
        # Seleciona uma extensão da lista, ciclicamente
        ext = extensoes[i % len(extensoes)]
        # Monta o nome do arquivo, ex: 'arquivo_teste_1.txt'
        nome_arquivo = f"arquivo_teste_{i}{ext}"
        # Monta o caminho completo para o novo arquivo dentro do diretório especificado
        caminho_arquivo = os.path.join(diretorio, nome_arquivo)
        # Abre o arquivo em modo de escrita ('w') e escreve um conteúdo simples
        with open(caminho_arquivo, 'w') as f:
            f.write(f"Conteúdo do {nome_arquivo}")
        # Informa que o arquivo foi criado
        print(f"  - Criado: {nome_arquivo}")

def organizar_arquivos_por_extensao(diretorio='.'):
    """Organiza arquivos em pastas baseadas em suas extensões.""" # Docstring explicando a função
    # Imprime o diretório que será organizado
    print(f"\nIniciando organização de arquivos em: {os.path.abspath(diretorio)}")
    # Itera sobre todos os itens (arquivos e pastas) no diretório especificado
    for nome_arquivo in os.listdir(diretorio):
        # Monta o caminho completo para cada item
        caminho_completo_arquivo = os.path.join(diretorio, nome_arquivo)

        # Verifica se o item é realmente um arquivo (e não uma pasta)
        if os.path.isfile(caminho_completo_arquivo):
            # Separa o nome do arquivo da sua extensão
            _, extensao = os.path.splitext(nome_arquivo)
            # Remove o ponto da extensão e a converte para minúsculas (ex: '.PDF' -> 'pdf')
            extensao = extensao.lower().strip('.')

            # Se o arquivo não tiver extensão (ex: 'meu_arquivo'), ele é ignorado
            if not extensao:
                continue

            # Define o nome da pasta de destino, que será a própria extensão (ex: 'pdf')
            pasta_destino = os.path.join(diretorio, extensao)
            # Cria a pasta de destino se ela não existir. exist_ok=True evita erro se já existir.
            os.makedirs(pasta_destino, exist_ok=True)

            # Move o arquivo para a pasta de destino correspondente à sua extensão
            shutil.move(caminho_completo_arquivo, os.path.join(pasta_destino, nome_arquivo))
            # Informa qual arquivo foi movido e para onde
            print(f"  - Movido '{nome_arquivo}' para '{pasta_destino}'")
    # Mensagem de conclusão da organização
    print("Organização concluída!")

# --- Execução do Programa ---
# Define o nome do diretório onde os arquivos de teste serão criados e organizados
diretorio_trabalho = 'arquivos_baguncados'
# Cria o diretório de trabalho se ele não existir
os.makedirs(diretorio_trabalho, exist_ok=True)

# 1. Chama a função para criar os arquivos de teste no diretório definido
criar_arquivos_teste(diretorio_trabalho)

# 2. Chama a função para organizar os arquivos criados
organizar_arquivos_por_extensao(diretorio_trabalho)
```


#**MISSÃO 02** CONSULTA SIMPLES A APIs


```python
'''
Desenvolver um script que consulte uma API pública (ViaCEP)
e retorne as informações de endereço formatadas de maneira legivel.
Integração com a API ViaCEP funcional. JSON recebido é formatado e impresso com clareza.
Erros de conexão ou CEP inválido são tratados sem 'quebrar' o script.
nao tire essa mensagem apenas adicione o codigo abaixo, coloque # para explicar cada linha
'''

import requests # Importa a biblioteca requests para fazer requisições HTTP a APIs
import json # Importa a biblioteca json para trabalhar com dados no formato JSON

def consultar_cep(cep):
    """Consulta a API ViaCEP para obter informações de endereço.""" # Docstring explicando a função
    # URL base da API ViaCEP, com um placeholder para o CEP
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # Faz a requisição GET para a URL da API
        response = requests.get(url, timeout=5) # Adiciona um timeout para evitar que a requisição demore indefinidamente
        # Verifica se a requisição foi bem-sucedida (código de status 200)
        response.raise_for_status() # Levanta uma exceção para códigos de status HTTP de erro (4xx ou 5xx)

        # Converte a resposta JSON em um dicionário Python
        dados_cep = response.json()

        # Verifica se o CEP é inválido (retorno da ViaCEP indica erro)
        if 'erro' in dados_cep and dados_cep['erro'] == True:
            print(f"Erro: CEP '{cep}' não encontrado ou inválido.")
            return None # Retorna None para indicar que a consulta falhou

        return dados_cep # Retorna o dicionário com os dados do endereço

    except requests.exceptions.HTTPError as http_err:
        # Captura erros de HTTP (ex: 404 Not Found, 500 Internal Server Error)
        print(f"Erro HTTP ao consultar o CEP '{cep}': {http_err}")
        return None
    except requests.exceptions.ConnectionError as conn_err:
        # Captura erros de conexão (ex: sem internet, DNS inválido)
        print(f"Erro de Conexão ao consultar o CEP '{cep}': {conn_err}")
        return None
    except requests.exceptions.Timeout as timeout_err:
        # Captura erros de timeout (requisição demorou demais)
        print(f"Tempo esgotado ao consultar o CEP '{cep}': {timeout_err}")
        return None
    except requests.exceptions.RequestException as req_err:
        # Captura outros erros gerais da biblioteca requests
        print(f"Ocorreu um erro inesperado na requisição para o CEP '{cep}': {req_err}")
        return None
    except json.JSONDecodeError:
        # Captura erro se a resposta não for um JSON válido
        print(f"Erro ao decodificar a resposta JSON da API para o CEP '{cep}'.")
        return None

def formatar_endereco(dados):
    """Formata as informações do endereço de forma legível.""" # Docstring explicando a função
    if not dados: # Se não houver dados, retorna uma mensagem de erro
        return "Não foi possível obter os dados do endereço."

    # Constrói uma string formatada com os detalhes do endereço
    endereco_formatado = f"Endereço Encontrado:\n"
    endereco_formatado += f"  CEP: {dados.get('cep', 'N/A')}\n"
    endereco_formatado += f"  Logradouro: {dados.get('logradouro', 'N/A')}\n"
    endereco_formatado += f"  Complemento: {dados.get('complemento', 'N/A')}\n"
    endereco_formatado += f"  Bairro: {dados.get('bairro', 'N/A')}\n"
    endereco_formatado += f"  Localidade: {dados.get('localidade', 'N/A')} - {dados.get('uf', 'N/A')}\n"
    endereco_formatado += f"  IBGE: {dados.get('ibge', 'N/A')}\n"
    endereco_formatado += f"  DDD: {dados.get('ddd', 'N/A')}"

    return endereco_formatado # Retorna a string formatada

# --- Exemplo de Uso --- #

# Solicita ao usuário que insira um CEP
cep_digitado = input("Digite um CEP para consulta (ex: 01001000): ")
# Remove espaços e hífens do CEP para garantir o formato correto para a API
cep_limpo = cep_digitado.strip().replace('-', '')

# Verifica se o CEP tem 8 dígitos, que é o formato esperado pela ViaCEP
if len(cep_limpo) == 8 and cep_limpo.isdigit():
    print(f"\nConsultando CEP: {cep_limpo}...")
    # Chama a função para consultar o CEP
    informacoes_endereco = consultar_cep(cep_limpo)

    # Se a consulta retornar dados, formata e imprime
    if informacoes_endereco:
        print(formatar_endereco(informacoes_endereco))
    else:
        # A mensagem de erro já foi impressa pela função 'consultar_cep'
        print("Consulta de CEP falhou.")
else:
    # Mensagem de erro para CEP com formato inválido antes mesmo de consultar a API
    print("Erro: CEP inválido. Por favor, digite 8 dígitos numéricos.")


```

```python
'''
Criar um programa autônomo que dispare alertas (mensagens de alerta no console ou e-mail simulado)
unicamente quando uma condição for atingida. Lógica condicional disparando corretamente.
Código modular (separado em funções). Refatoração visível entre a primeira versão e a versão final.
nao tire essa mensagem apenas adicione o codigo abaixo, coloque # para explicar cada linha

'''

# --- VERSÃO INICIAL --- #

def verificar_condicao_inicial(valor_monitorado):
    # Define uma condição simples: se o valor_monitorado for maior que 100
    if valor_monitorado > 100:
        return True # Retorna True se a condição for atingida
    else:
        return False # Retorna False caso contrário

def disparar_alerta_console(mensagem):
    # Função para simular o disparo de um alerta no console
    print(f"ALERTA (CONSOLE): {mensagem}") # Imprime a mensagem de alerta no console

def programa_autonomo_v1(dado):
    # Função principal do programa autônomo (Versão 1)
    print(f"Monitorando dado: {dado}") # Informa qual dado está sendo monitorado
    # Verifica se a condição é atingida chamando a função de verificação
    if verificar_condicao_inicial(dado):
        # Se a condição for verdadeira, dispara o alerta
        disparar_alerta_console(f"Condição crítica atingida! Valor: {dado}")
    else:
        # Se a condição não for verdadeira, informa que está tudo ok
        print("Condição normal. Nenhum alerta disparado.")

# --- Exemplo de Uso da Versão Inicial --- #
print("\n--- Executando Programa Autônomo V1 ---")
programa_autonomo_v1(80) # Teste com um valor que não atinge a condição
programa_autonomo_v1(120) # Teste com um valor que atinge a condição
programa_autonomo_v1(95) # Outro teste com um valor que não atinge
```
