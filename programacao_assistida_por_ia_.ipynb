```python
numeros = [1, 2, 3, 4, 5] #lista de numeros
soma = 0 # vai começar aqui
i = 0 # i vai ser a posição da lista
while i < len(numeros):# vai ver se i e menor que a quantidade de items na lista e criar um loop
  soma += numeros[i]# soma vai aderir soma + um indice da lista
  i += 1 # vai somar mais 1 na variavel i

  print(soma) # vai printar na tela
```
